## ハフマン符号

### 符号化
* 以下のコードで，ハフマン符号の符号化ができる
* 情報源（`source`）を指定し，`huffman(source)`を実行することで符号化が行われる
* `source`の定義:
  * `source = {"a": 確率, "b": 確率, "c": 確率, ...}`の形式で記述
  * `"情報源記号": 確率` 前後の改行や空白は任意に入れられる（入れなくてもよい）
  * `"情報源記号": 確率` は，カンマで区切る
* アルゴリズム（同確率の場合の優先ルールなど）は講義資料と同じ

```python
source = {
    "a": 0.55,
    "b": 0.02,
    "c": 0.03,
    "d": 0.25,
    "e": 0.15
}

huffman(source)
```
