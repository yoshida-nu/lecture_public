## UTF-8

### 符号化
* 以下のコードで，Unicodeの文字（正確には番号）を UTF-8 で符号化した結果（符号語と呼ぶ）を確認する
* `utf8_hex("文字")`：
  * 指定した文字を UTF-8 で符号化した符号語を出力する
  * 文字は 1文字だけ指定する
  * `utf8_hex`の下に赤い波線が表示されるかもしれないが，正常に実行できていれば問題ないので気にしなくてよい
* 結果から，文字によって，符号語の長さ（16進数の桁数）が異なることがわかる
* 16進数1桁が4ビットなので，「A」は8ビット（1バイト），「あ」は24ビット（3バイト）となる ⇒ UTF-8はバイト列に符号化している
* ASCIIコードで扱っている文字（半角英数字や記号）は，Unicodeの番号と符号化されたバイト列が一致する
* UTF-8では，ASCIIコードの7ビットを8ビットで表現している ⇒ 最上位ビットが 0 になる

```python
utf8_hex("A")
utf8_hex("あ")
```

# コンピュータ実習（２）

## 音の表現
* 本実習では，サンプリング周波数と量子化ビット数の違いで，音の波形や聴こえ方がどのように変わるのかを確認する

### 全体の流れ
* ここで用いるコードは，次を繰り返し体験させるためのもの：
1. 基準音（元の波形）を作る
2. サンプリング周波数を変えて標本化（正確にはリサンプリング）する ⇒ サンプリング周波数の違いを作る
3. 量子化ビット数を変えて量子化する ⇒ 量子化ビット数の違いを作る
4. 波形をグラフに描画する ⇒ 見た目を比較
5. 音声ファイル（WAVファイル）に保存する ⇒ 聴こえ方を比較
* `run_experiment(fs_list, bits_list)`:
  * `fs_list`: サンプリング周波数（Hz）のリスト（カンマで区切って複数指定できる）
  * `bits_list`: 量子化ビット数のリスト（カンマで区切って複数指定できる）
  * `fs_list` と `bits_list` のすべての組み合わせに対する波形のグラフを描画
  * 同様に `fs_list` と `bits_list` のすべての組み合わせに対する音声ファイルを作成
  * 音声ファイルは，ノートブック上にフォルダ「out_wav」が自動的に作成され，その中に保存される
  * ファイル名の上をダブルクリックするとダウンロードできる
  * `run_experiment`の下に赤い波線が表示されるかもしれないが，正常に実行できていれば問題ないので気にしなくてよい

```python
run_experiment([4000], [4])
```

### 実習
* `run_experiment` を使って，いくつかのサンプリング周波数と量子化ビット数の組み合わせで，波形（見た目）と音声（聴こえ方）を比較する
* PCに接続できるイヤホンがなければ，波形だけを比較する
* 特に，以下について検討する
  * サンプリング周波数を下げると，波形はどのように変わるか？ （なめらかさ，点の間隔など）
  * サンプリング周波数を下げると，音はどのように変わるか？ （こもる，高音が消える など）
  * 量子化ビット数を下げると，波形はどう変わるか？
  * 量子化ビット数を下げると，音はどう変わるか？
