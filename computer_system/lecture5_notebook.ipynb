## 加算器の原理
* 以下のコードで，AND，XOR，シフト演算によるビット列の加算の処理過程を観察する
* `add_by_logic("a", "b")`:
  * `a`, `b`: ビット列
  * `add_by_logic("a", "b")`は，2つのビット列 `a` と `b` の加算を AND，XOR，シフト演算を使って行っている過程を出力する
  * XOR: 桁上りを無視した加算 ⇒ XORで計算
  * CAR: 桁上りした数（CAR は Carry の略）⇒ AND と 1ビット左シフトで計算
  * XORとCARの加算が `a`と`b`の加算に一致する
  * `add_by_logic`の下に赤い波線が表示されるかもしれないが，正常に実行できていれば問題ないので気にしなくてよい

```python
add_by_logic("101", "011")
```
