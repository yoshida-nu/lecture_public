# オリジナルのシーザー暗号

## 1文字だけシーザー暗号
* まずは，1文字だけ暗号化できるシーザー暗号（鍵は3に固定）のコードを作成する

### 1文字だけ暗号化
* `ord`関数と`chr`関数を利用する
* 以下のコードは，暗号化する文字を数値に変換する⇒数値を3つずらす⇒文字（暗号文）に戻すの順に処理を進めることで，1文字だけの暗号化を実現している
* `print(ciphertext.upper())`: `ciphertext` を大文字にして出力

````python
plaintext = "a"
number = ord(plaintext)
number = number + 3
ciphertext = chr(number)
print(ciphertext.upper())
````

* 上のコードは，まだ不完全
* 試しに，上のコードの「a」を「y」に変更して実行結果を確認してみる
* 「y」に対する暗号文は「b」となるが，そのような結果にならない
* 今のコードは単にUnicodeの番号に3を足しているだけなので，「z」以降の文字になっている
* 正確な暗号化処理にするには，アルファベット26文字で循環させる必要がある ⇒ 26で割った余りをとる
* 以下のコードが，完全な1文字だけ暗号化となる
  * `number`が 0～25（a～zに対応）の値をとるように，Unicodeの番号を変換 ⇒ `number = ord(plain) - ord("a")`
  * `number = (number + 3) % 26`で暗号化（26文字で循環）
  * この時点で`number`は，Unicodeの番号に対応してないので，Unicodeの番号に変換して文字に戻す ⇒ `chr(number + ord("a"))`

````python
plaintext = "a"
number = ord(plaintext) - ord("a")
number = (number + 3) % 26
ciphertext = chr(number + ord("a"))
print(ciphertext.upper())
````

## `for`文の復習
*  `for`の直後に半角スペースを入れたあと，繰り返しの処理の中で用いる変数を記述
*  その変数の直後に半角スペースを入れたあと，`in`と記述
*  `in`の直後に半角スペースを入れたあと，特定の範囲の数値を指定できる`range`やリストを記述し，最後にコロン`:`を入れる
*  各繰り返しで，`in`の直後で指定した`range`やリストの要素が順番に1つずつ変数に格納される
*  取り出す要素がなくなったら繰り返しを終了する

---
書式:  
```python
for 変数 in rangeやリスト:
	処理
```
---

* 実は，`in`の直後に文字列を指定することもできる
* 文字列の場合は，先頭の文字から 1文字ずつ取り出される
* 以下のコードは，文字列「python」を先頭から1文字ずつ取り出して，出力している
* `print(i.upper(), end="")`: `i` のうしろに空白を入れないで，大文字にして出力

````python
text = "python"
for i in text:
	print(i.upper(), end="")
````

# 一般のシーザー暗号

* 鍵の値を格納するための変数`key`を用いで一般のシーザー暗号の暗号化と復号を実現する

## 一般のシーザー暗号の暗号化

````python
key = 5
plaintext = "python"
for i in plaintext:
	number = ord(i) - ord("a")
	number = (number + key) % 26
	ciphertext = chr(number + ord("a"))
	print(ciphertext.upper(), end="")
````
