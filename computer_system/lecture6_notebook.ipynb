## 命令実行サイクル

### 本実習における命令の種類と構造
* 本実習は，命令実行サイクルのイメージをつかむことが目的なので，次の4つの命令のみを扱う
  * `LOAD`： 転送命令 ⇒ レジスタに格納
  * `ADD`： 算術演算命令 ⇒ レジスタのデータと加算して上書き
  * `SUB`： 算術演算命令 ⇒ レジスタのデータと減算して上書き
  * `JUMP`： 分岐命令 ⇒ 指定されたプログラムカウンタに移動
* また，メモリに格納されている命令は，命令と実際のデータが格納されているものとする（オペランド読み出しを省略）

### 本実習の狙い
* 実習内容： Pythonを使って，CPU内での動作を擬似的に体験する
* 以下の処理が順番に行われて，命令が実行されていること説明（言語化）できるようにする
  * CPUがメモリに入っている命令を一つずつ取り出す
  * レジスタが値を保持する
  * 命令実行の流れ（フェッチ ⇒ デコード ⇒ 実行）
* 例として，5 + 3 - 2 を計算する処理を考える
* 答えが「6」であることは簡単にわかるが，本実習では「答え」ではなく「過程」を見る


### コードについて
* 以下のコードで，CPUにおける簡易的な命令実行プロセスを観察する
* `memory = [ ... ]`部分がメモリに入っている命令に相当する
> <img src="./fig/code_and_memory.jpg" width="450">
* `run_program(memory)`:
  * `memory`: メモリ内の命令
  * `run_program(memory)`は，メモリ内の命令 `memory` を CPU が実行する処理過程を出力する
  * `run_program`の下に赤い波線が表示されるかもしれないが，正常に実行できていれば問題ないので気にしなくてよい

````python
memory = [
    ("LOAD", 5),
    ("ADD", 3),
    ("SUB", 2),
]

run_program(memory)
````

#### 命令実行（1サイクル目）
```code
====1回目のサイクル====
プログラムカウンタ: 0
命令: LOAD
データ: 5
現在のレジスタ: 5
```
* フェッチ（命令取り出し）：
  * プログラムカウンタ（PC）内のアドレスから命令を取り出す ⇒ ('LOAD', 5)
  * プログラムカウンタを1つ繰り上げる→ 0 ⇒ 1
* デコード（命令解読）： 命令部 'LOAD' ⇒ レジスタにデータを格納
* 実行： レジスタに「5」を格納 → 0 ⇒ 5

> <img src="./fig/command_cycle1.jpg" width="550">

#### 命令実行（2サイクル目）
```code
====2回目のサイクル====
プログラムカウンタ: 1
命令: ADD
データ: 3
現在のレジスタ: 8
```
* フェッチ（命令取り出し）：
  * プログラムカウンタ内のアドレスから命令を取り出す ⇒ ('ADD', 3)
  * プログラムカウンタを1つ繰り上げる→ 1 ⇒ 2
* デコード： 命令部 'ADD' ⇒ レジスタの値「5」とデータ「3」を加算
* 実行： レジスタに加算した結果「8」を格納 → 5 ⇒ 8

> <img src="./fig/command_cycle2.jpg" width="550">

#### 命令実行（3サイクル目）
```code
====3回目のサイクル====
プログラムカウンタ: 2
命令: SUB
データ: 2
現在のレジスタ: 6
```
* フェッチ（命令取り出し）：
  * プログラムカウンタ内のアドレスから命令を取り出す ⇒ ('SUB', 2)
  * プログラムカウンタを1つ繰り上げる→ 2 ⇒ 3
* デコード： 命令部 'SUB' ⇒ レジスタの値「8」からデータ「2」を減算する
* 実行： レジスタに減算した結果「6」を格納 → 8 ⇒ 6

> <img src="./fig/command_cycle3.jpg" width="550">

#### 実行完了
```code
====== 実行完了 ======
最終結果: 6
```
* すべての命令を実行したので，処理完了
* この時点でレジスタに入っている内容が最終結果（実行結果）となる

## CPU の性能評価
* 以下のコードで，CPU の性能評価指標を自動計算する
* `cpu_performance_report(value, "aux", CPI)`:
  * `value`: クロック周波数（数値のみ）
  * `aux`: 補助単位（SI接頭語）
    * なし: `""`
    * キロ: `"K"`
    * メガ: `"M"`
    * ギガ: `"G"`
  * `CPI`: CPI（1命令を実行するために必要なクロック数）
  * `cpu_performance_report`を呼び出すと，引数で指定した CPU の性能評価指標等を出力する
  * `2e+09` は $2 \times 10^9$ を意味する
  * `cpu_performance_report`の下に赤い波線が表示されるかもしれないが，正常に実行できていれば問題ないので気にしなくてよい
* 以下のコードでは，クロック周波数 2GHz, 8CPIで動作するCPUの性能評価結果を出力している

````python
cpu_performance_report(2, "G", 8)
````
