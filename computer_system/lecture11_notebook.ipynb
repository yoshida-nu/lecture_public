# コンピュータ実習（１） ーワンタイムパッド暗号ー
* 暗号化と復号は，ビット列に対する排他的論理和（XOR）なので，`bit_op`関数を用いればよい（[第5回実習用ノートブック](https://colab.research.google.com/github/yoshida-nu/lecture_public/blob/main/computer_system/lecture5_notebook.ipynb)参照）
* `bit_op("a", "b", "op")`:
  * `a`, `b`: ビット列
  * `op`: 演算の種類（AND, OR, XORなど）
  * `bit_op("a", "b", "op")`を呼び出すと，2つのビット列 `a` と `b` を `op` で演算した結果を戻り値として返す
  * `bit_op`の下に赤い波線が表示されるかもしれないが，正常に実行できていれば問題ないので気にしなくてよい

## 暗号化の例
* 次のコードでは，平文「10011」を鍵「11011」で暗号化した暗号文を出力している

```python
plaintext = "10011"
key = "11011"
ciphertext = bit_op(plaintext, key, "XOR")
print(ciphertext)
```

## 実習：ワンタイムパッド暗号の暗号化
* 以下のコードセルに，平文「011」を鍵「110」で暗号化した暗号文を出力するコードを作成し，実行せよ

```python
plaintext = "011"
key = "110"
ciphertext = bit_op(plaintext, key, "XOR")
print(ciphertext)
```

## 復号の例
* 次のコードでは，暗号文「01000」を鍵「11011」で復号した復号文を出力している
* 演算は，暗号化と同じ

```python
ciphertext = "01000"
key = "11011"
plaintext = bit_op(ciphertext, key, "XOR")
print(plaintext)
```
