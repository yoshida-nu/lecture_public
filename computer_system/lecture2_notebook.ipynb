## 桁あふれ誤差
* 以下のコードで，桁あふれ誤差を擬似的に体験する
* 意図しない値になっていることが確認できればよい
* `signed_nbit_repr(x, n, overflow=True)`：
  * `x`: 整数（10進数）
  * `n`: ビット数
  * 整数 `x` を `n`ビットの符号付き整数として表現したビット列を出力する
  * `overflow=True`を記述する: `x` が $-2^{n-1}$ ～ $2^{n-1}-1$の範囲外の場合，桁あふれ誤差が発生（`x` を $2^n$ で割った余りを基数変換）した状態でビット列を出力する
  * `overflow=True`を記述しない: `x` が範囲外の場合はエラー表示となる
  * `show_signed_nbit_repr`の下に赤い波線が表示されるかもしれないが気にしなくてよい


```python
show_signed_nbit_repr(130, 8, overflow=True)
show_signed_nbit_repr(-129, 8, overflow=True)
```

## 実習
* 以下のコードセルに，整数「8」と「-9」をそれぞれ 4ビットの符号付き整数として表現したビット列を出力するコードを作成し，実行せよ
* 「8」と「-9」が $-2^{3}$ ～ $2^{3}-1 = -8$ ～ $7$の範囲外なので桁あふれ誤差が発生することが確認できる
